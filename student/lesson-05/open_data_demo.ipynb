```python
import pandas as pd
import io
import urllib.request

pd.set_option("display.max_columns", None)
```

## 1. Ingest

Edmonton's portal runs on **Socrata**, which exposes every dataset as a simple web API (SODA). Same data, two formats — this is the CSV vs. JSON distinction from today's slides:

- CSV: `https://data.edmonton.ca/resource/eecg-fc54.csv?$limit=1000`
- JSON: `https://data.edmonton.ca/resource/eecg-fc54.json?$limit=1000`

`$limit` caps how many rows come back — useful manners when a dataset has 300k+ rows and you only want a look.

If the classroom wifi is uncooperative, the fallback cell below builds a small **illustrative** sample with the same columns, so the demo still runs offline. It is not real City of Edmonton data — just a stand-in with the right shape.

```python
CSV_URL = "https://data.edmonton.ca/resource/eecg-fc54.csv?$limit=1000"

FALLBACK_CSV = """id,genus,species_botanical,species_2,location_type,bears_edible_fruit,condition_percent,location_zip,latitude,longitude
100234,Malus,Malus domestica,Apple,Boulevard,Yes,82,T5J,53.5461,-113.4938
100235,Acer,Acer saccharinum,Silver Maple,Park,No,71,T6G,53.5232,-113.5263
100236,Fraxinus,Fraxinus pennsylvanica,Green Ash,Boulevard,No,,T5N,53.5591,-113.5102
100237,Malus,malus domestica,apple,boulevard,yes,90,T5J,53.5478,-113.4951
100238,Picea,Picea glauca,White Spruce,Park,No,65,T6E,53.5024,-113.4783
100239,Prunus,Prunus cerasus,Sour Cherry,Boulevard,Yes,88,T6H,53.4890,-113.4771
100240,Acer,Acer saccharinum,Silver Maple,Park,No,40,T6G,53.5240,-113.5271
100241,Fraxinus,Fraxinus pennsylvanica,Green Ash,Boulevard,No,74,,53.5605,-113.5119
100242,Malus,Malus domestica,Apple,Boulevard,Yes,77,T5J,53.5465,-113.4940
100243,Ulmus,Ulmus americana,American Elm,Park,No,95,T6C,53.5401,-113.4623
"""

try:
    urllib.request.urlopen(CSV_URL, timeout=5)
    df = pd.read_csv(CSV_URL)
    source = "live portal"
except Exception as e:
    print(f"Couldn't reach the portal ({e!r}) — using the offline fallback sample instead.")
    df = pd.read_csv(io.StringIO(FALLBACK_CSV))
    source = "offline fallback sample"

print(f"Loaded {len(df)} rows from: {source}")
```

## 3. Same data, other format (CSV vs. JSON)

`pandas` reads both the same way — `read_csv` vs `read_json`. Worth showing once so the format talk from the slides isn't abstract.

```python
JSON_URL = "https://data.edmonton.ca/resource/eecg-fc54.json?$limit=5"

try:
    urllib.request.urlopen(JSON_URL, timeout=5)
    df_json = pd.read_json(JSON_URL)
    display(df_json.head())
except Exception as e:
    print(f"Skipping the live JSON pull ({e!r}) — same idea as above, just a different file format.")
```

## 4. First exposure to messy data

Before computing anything, check what's missing or inconsistent — exactly what the Dataset Evaluation Checklist calls **completeness** and **consistency**.

```python
# Missing values per column
df.isna().sum()
```

```python
# Naming/capitalization inconsistency is common in open data ("Apple" vs "apple")
if "species_2" in df.columns:
    print(df["species_2"].str.lower().value_counts())
```

## 5. Summary stats — the same numbers, in code

This mirrors the `=AVERAGE()` / `=MEDIAN()` / `=MODE()` formulas from the spreadsheet activity, on the tree condition score (0–100).

```python
col = "condition_percent"
if col in df.columns:
    print("Mean:  ", df[col].mean())
    print("Median:", df[col].median())
    print("Mode:  ", df[col].mode().iloc[0])
    print("Std dev:", df[col].std())
    print("IQR:   ", df[col].quantile(0.75) - df[col].quantile(0.25))
    df[col].describe()
```

## 6. One picture is worth a lot of formulas

A histogram shows shape and outliers at a glance — the same concepts from the Connecting warm-up.

```python
import matplotlib.pyplot as plt

if col in df.columns:
    df[col].plot(kind="hist", bins=10, edgecolor="white")
    plt.title("Distribution of tree condition score")
    plt.xlabel("Condition (%)")
    plt.show()
```
